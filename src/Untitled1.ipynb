```python
movies = spark.read.csv(path, header=True, inferSchema=True)
crew = spark.read.csv("hdfs://namenode:9000/data/crew.csv", header=True, inferSchema=True)
genre = spark.read.csv("hdfs://namenode:9000/data/genres.csv", header=True, inferSchema=True)
language = spark.read.csv("hdfs://namenode:9000/data/languages.csv", header=True, inferSchema=True)
actors = spark.read.csv("hdfs://namenode:9000/data/actors.csv", header=True, inferSchema=True)
releases = spark.read.csv("hdfs://namenode:9000/data/releases.csv", header=True, inferSchema=True)
studios = spark.read.csv("hdfs://namenode:9000/data/studios.csv", header=True, inferSchema=True)
themes = spark.read.csv("hdfs://namenode:9000/data/themes.csv", header=True, inferSchema=True)

```

```python
movies.printSchema()
crew.printSchema()
genre.printSchema()
language.printSchema()
actors.printSchema()
releases.printSchema()
studios.printSchema()
themes.printSchema()
```

```python
from pyspark.sql.functions import col

# movies
m = movies.select(
    col("id"),
    col("name").alias("movie_name"),
    col("date").alias("movie_date"),
    col("tagline").alias("movie_tagline"),
    col("description").alias("movie_description"),
    col("minute").alias("movie_minute"),
    col("rating").alias("movie_rating")
)

# crew
c = crew.select(
    col("id"),
    col("name").alias("crew_name"),
    col("role").alias("crew_role")
)

# genre
g = genre.select(
    col("id"),
    col("genre").alias("genre_name")
)
l = language.select(
    col("id"),
    col("type").alias("languages_type"),
    col("language").alias("languages_languages")

)
ac = actors.select(
    col("id"),
    col("name").alias("actors_name"),
    col("role").alias("actors_role")
)

r = releases.select(
    col("id"),
    col("country"),
    col("date").alias("release_date"),
    col("type").alias("release_type"),
    col("rating").alias("release_rating")
)

s = studios.select(
    col("id"),
    col("studio")
)

t = themes.select(
    col("id"),
    col("theme")
)
```

```python
df = (m.join(g, "id", "inner") \
      .join(c, "id", "inner")
      .join(l, "id", "inner")
      .join(ac,"id", "inner")
      .join(r, "id", "inner")
      .join(s, "id", "inner")
      .join(t, "id", "inner")
)
```

```python
df_filtered = df.select("movie_name", "genre_name", "crew_name", "crew_role")
df_filtered.show(10,truncate=False)

```

```python
df_filtered.filter(col("genre_name") == "Comedy").show(truncate=False)


```

```python
from pyspark.sql.functions import col

df.filter(col("crew_name").contains("Nolan")) \
  .select("movie_name", "genre_name", "crew_role") \
  .show(truncate=False)

```

```python


df.withColumn("movie_rating", col("movie_rating").cast("double")) \
  .orderBy(col("movie_rating").desc()) \
  .select("movie_name", "movie_rating", "genre_name") \
  .show(10, truncate=False)
```

```python
df.groupBy("genre_name") \
  .count() \
  .orderBy(col("count").desc()) \
  .show(truncate=False)
```
