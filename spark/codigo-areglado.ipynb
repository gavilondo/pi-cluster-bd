```python
from pyspark.sql import functions as F

# --- PASO 0: CARGA REAL DE DATOS ---
# (Aquí es donde definimos las variables que te daban error)
df_mov = spark.read.csv("hdfs://namenode:9000/data/archive/movies.csv", header=True, inferSchema=True)
df_lan = spark.read.csv("hdfs://namenode:9000/data/archive/languages.csv", header=True, inferSchema=True)
df_gen = spark.read.csv("hdfs://namenode:9000/data/archive/genres.csv", header=True, inferSchema=True)
df_act = spark.read.csv("hdfs://namenode:9000/data/archive/actors.csv", header=True, inferSchema=True)
df_cou = spark.read.csv("hdfs://namenode:9000/data/archive/countries.csv", header=True, inferSchema=True)
df_cre = spark.read.csv("hdfs://namenode:9000/data/archive/crew.csv", header=True, inferSchema=True)
df_rel = spark.read.csv("hdfs://namenode:9000/data/archive/releases.csv", header=True, inferSchema=True)
df_std = spark.read.csv("hdfs://namenode:9000/data/archive/studios.csv", header=True, inferSchema=True)
df_thm = spark.read.csv("hdfs://namenode:9000/data/archive/themes.csv", header=True, inferSchema=True)

# --- PASO 1: Ahora sí, definimos el diccionario ---
processed_dfs = {
    "mov": df_mov,
    "lan": df_lan,
    "gen": df_gen,
    "act": df_act,
    "cou": df_cou,
    "cre": df_cre,
    "rel": df_rel,
    "std": df_std,
    "thm": df_thm
}

# --- PASO 2: Configuración de columnas ---
config_tablas = {
    "lan": "language",
    "gen": "genre",
    "act": "name",      
    "cou": "country",
    "cre": "role",      
    "rel": "date",      
    "std": "studio",
    "thm": "theme"
}

# --- PASO 3: Agrupación y Join (El resto de tu código) ---
tablas_agrupadas = {}
for alias, col_nombre in config_tablas.items():
    tablas_agrupadas[alias] = processed_dfs[alias] \
        .groupBy("id") \
        .agg(F.collect_set(col_nombre).alias(f"list_{alias}"))

final_df = processed_dfs["mov"]
for alias in config_tablas.keys():
    final_df = final_df.join(tablas_agrupadas[alias], "id", "left")

# --- PASO 4: Selección y Vista Temporal ---
resultado = final_df.select(
    F.col("id"),
    F.col("name").alias("movie_title"),
    F.col("date").alias("release_date"),
    F.col("minute").cast("int"),
    F.col("rating").cast("float"),
    F.col("list_gen").alias("genres"),
    F.col("list_lan").alias("languages"),
    F.col("list_act").alias("actors")
).orderBy("id")

resultado.createOrReplaceTempView("tabla_cine")
resultado.show(5)
```
